```python
dicionario = {}

def verificar_senhas_repetidas(dic):
    """Verifica se há senhas repetidas entre serviços diferentes."""
    senhas_servicos = {}
    for servico, senha in dic.items():
        if senha in senhas_servicos:
            senhas_servicos[senha].append(servico)
        else:
            senhas_servicos[senha] = [servico]

    tem_risco = False
    print("\n--- ANÁLISE DE RISCO (Senhas Repetidas) ---")
    for senha, servicos in senhas_servicos.items():
        if len(servicos) > 1:
            tem_risco = True
            print(f" Alerta de Risco! A mesma senha está sendo usada nos seguintes serviços: {', '.join(servicos)}")

    if not tem_risco:
        print(" Nenhuma senha repetida encontrada. Ótimo trabalho!")

print(" - Menu:    1: Cadastrar - 2: Consultar - 3: Atualizar - 4: Excluir - 5: Verificar Riscos - 0: Sair ")

while True:
    escolha = int(input("\nDigite a opção desejada: "))

    if escolha == 1:
        cadastro = input("Escreva o aplicativo (serviço): ")
        nome = input("Insira a senha desse aplicativo: ")
        dicionario[cadastro] = nome
        print("Senha cadastrada com sucesso!")

    elif escolha == 2:
        consulta = input("Digite o aplicativo: ")
        print(dicionario.get(consulta, "codigo invalido!"))

    elif escolha == 3:
        atualizar = input("Digite o aplicativo: ")
        if atualizar in dicionario:
            novo_nome = input("Digite a nova senha: ")
            dicionario[atualizar] = novo_nome
            print("Senha atualizada!")
        else:
            print("Aplicativo não encontrado.")

    elif escolha == 4:
        excluir = input("Digite o aplicativo que deseja pra excluir a senha: ")
        if excluir in dicionario:
            del(dicionario[excluir])
            print("Excluído com sucesso!")
        else:
            print("Aplicativo não encontrado.")

    elif escolha == 5:
        verificar_senhas_repetidas(dicionario)

    elif escolha == 0:
        break

    else:
        print("Opção invalida!")
```
