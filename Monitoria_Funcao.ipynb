```python
def substituir_espacos(texto, texto_novo):
  textos = []
  textos.append(texto)
  for i in range(len(textos)):
    if texto[i] == " ":
      texto_novo.append(textos)

  return textos
texto = str(input("Dgite o texto: "))
caractere = str(input("Digite o caractere: "))

imprimir = substituir_espacos(texto, caractere)
print(imprimir)
```

```python
def separar_sinais(lista_numeros):
    positivos = []
    negativos = []

    for numero in lista_numeros:
        if numero >= 0:
            positivos.append(numero)
        else:
            negativos.append(numero)

    return positivos, negativos

# --- PARTE DO INPUT ---

# 1. Pedimos ao utilizador para digitar os números separados por espaço
entrada = input("Digite os números separados por espaço (ex: 10 -3 0 5): ")

# 2. Dividimos a string em pedaços e convertemos cada um para inteiro
# O método .split() separa o texto onde houver espaços
lista_usuario = [int(x) for x in entrada.split()]

# 3. Chamamos a nossa função com a lista criada pelo utilizador
pos, neg = separar_sinais(lista_usuario)

# 4. Mostramos os resultados no ecrã
print("Positivos (e zero):", pos)
print("Negativos:", neg)
```

```python
def separar_sinais(lista_numeros):
    positivos = []
    negativos = []

    for numero in lista_numeros:
        if numero >= 0:
            positivos.append(numero)
        else:
            negativos.append(numero)

    return positivos, negativos

# --- PARTE DO INPUT COM WHILE TRUE ---

lista_usuario = []

print("Digite os números um a um. Quando terminar, digite 'sair' ou aperte Enter sem digitar nada.")

while True:
    entrada = input("Digite um número: ")

    # Condição para parar o ciclo (se o utilizador não digitar nada ou escrever 'sair')
    if entrada == "" or entrada.lower() == "sair":
        break

    # Tentamos converter para inteiro e adicionar à lista
    try:
        numero = int(entrada)
        lista_usuario.append(numero)
    except ValueError:
        print("Por favor, digite um número inteiro válido!")

# Chamamos a nossa função com a lista que o utilizador criou
pos, neg = separar_sinais(lista_usuario)

# Mostramos os resultados no ecrã
print("\n--- Resultados ---")
print("Positivos (e zero):", pos)
print("Negativos:", neg)
```
